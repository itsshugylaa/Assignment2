```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

data_dir = '/content/drive/MyDrive/Household_Dataset'
```

```python
img_height = 224
img_width = 224
batch_size = 32

print("Загрузка обучающей выборки:")
train_ds = tf.keras.utils.image_dataset_from_directory(
  data_dir,
  validation_split=0.2,
  subset="training",
  seed=123,
  image_size=(img_height, img_width),
  batch_size=batch_size)

print("Загрузка валидационной выборки:")
val_ds = tf.keras.utils.image_dataset_from_directory(
  data_dir,
  validation_split=0.2,
  subset="validation",
  seed=123,
  image_size=(img_height, img_width),
  batch_size=batch_size)

class_names = train_ds.class_names
print("Найдены классы:", class_names)
```

```python
data_augmentation = keras.Sequential(
  [
    layers.Rescaling(1./255, input_shape=(img_height, img_width, 3)),
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),
  ]
)
print("Слои аугментации и нормализации успешно созданы!")
```

```python
data_augmentation = keras.Sequential(
  [
    keras.Input(shape=(img_height, img_width, 3)),
    layers.Rescaling(1./255),
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),
  ]
)

num_classes = len(class_names)

baseline_model = keras.Sequential([
  data_augmentation,
  layers.Conv2D(16, 3, padding='same', activation='relu'),
  layers.MaxPooling2D(),
  layers.Conv2D(32, 3, padding='same', activation='relu'),
  layers.MaxPooling2D(),
  layers.Flatten(),
  layers.Dense(64, activation='relu'),
  layers.Dense(num_classes)
])

baseline_model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])

baseline_model.summary()
```

```python
epochs = 10

print("Начинаем обучение базовой модели...")
history = baseline_model.fit(
  train_ds,
  validation_data=val_ds,
  epochs=epochs
)
```

```python
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(img_height, img_width, 3),
    include_top=False,
    weights='imagenet'
)
base_model.trainable = False

improved_model = tf.keras.Sequential([
  keras.Input(shape=(img_height, img_width, 3)),
  layers.Lambda(tf.keras.applications.mobilenet_v2.preprocess_input),
  layers.RandomFlip("horizontal"),
  layers.RandomRotation(0.1),

  base_model,

  layers.GlobalAveragePooling2D(),
  layers.Dense(128, activation='relu'),
  layers.Dropout(0.2),
  layers.Dense(num_classes)
])

improved_model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])

print("Transfer Learning...")
improved_history = improved_model.fit(
  train_ds,
  validation_data=val_ds,
  epochs=10
)
```

```python
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

print("Сбор данных для оценки...")
val_images = []
val_labels = []
for images, labels in val_ds.unbatch():
    val_images.append(images.numpy())
    val_labels.append(labels.numpy())

val_images = np.array(val_images)
val_labels = np.array(val_labels)

print("Генерация предсказаний...")
predictions = improved_model.predict(val_images)
predicted_classes = np.argmax(predictions, axis=1)

print("\n--- Classification Report ---")
print(classification_report(val_labels, predicted_classes, target_names=class_names))

cm = confusion_matrix(val_labels, predicted_classes)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Предсказанный класс (Что сказала модель)')
plt.ylabel('Истинный класс (Как было на самом деле)')
plt.title('Confusion Matrix')
plt.show()
```

```python
!pip install gradio -q

import gradio as gr
import numpy as np
import tensorflow as tf
import cv2

class_names = ['bottles', 'cups', 'keys']

def predict_and_explain(image):
    if image is None:
        return "Пожалуйста, загрузите изображение.", None

    try:
        if len(image.shape) == 3 and image.shape[2] == 4:
            image = image[:, :, :3]
        image = np.array(image, dtype=np.uint8)

        img_resized = cv2.resize(image, (224, 224))
        img_array = tf.expand_dims(img_resized, 0)
        img_array = tf.cast(img_array, tf.float32)

        predictions = improved_model.predict(img_array)
        score = tf.nn.softmax(predictions[0])
        class_idx = np.argmax(score)
        predicted_class = class_names[class_idx]
        confidence = 100 * np.max(score)

        result_text = f"Результат: {predicted_class} (Уверенность: {confidence:.2f}%)"

        base_model = None
        for layer in improved_model.layers:
            if isinstance(layer, tf.keras.Model):
                base_model = layer
                break

        if base_model is None:
            return result_text + " (Base model not found)", image

        last_conv_layer = base_model.get_layer("out_relu")
        grad_model = tf.keras.models.Model(
            [base_model.inputs], [last_conv_layer.output, base_model.output]
        )

        with tf.GradientTape() as tape:
            img_preprocessed = tf.keras.applications.mobilenet_v2.preprocess_input(tf.identity(img_array))
            conv_outputs, preds = grad_model(img_preprocessed)
            loss = preds[:, class_idx]

        grads = tape.gradient(loss, conv_outputs)
        pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))

        conv_outputs = conv_outputs[0]
        heatmap = conv_outputs @ pooled_grads[..., tf.newaxis]
        heatmap = tf.squeeze(heatmap)
        heatmap = tf.maximum(heatmap, 0) / tf.math.reduce_max(heatmap)
        heatmap = heatmap.numpy()

        heatmap_uint8 = np.uint8(255 * heatmap)
        heatmap_colored = cv2.applyColorMap(heatmap_uint8, cv2.COLORMAP_JET)
        heatmap_resized = cv2.resize(heatmap_colored, (image.shape[1], image.shape[0]))

        heatmap_resized = np.array(heatmap_resized, dtype=np.uint8)
        superimposed_img = cv2.addWeighted(image, 0.6, heatmap_resized, 0.4, 0)

        return result_text, superimposed_img

    except Exception as e:
        print(f"Техническая ошибка внутри функции: {e}")
        fallback_text = result_text if 'result_text' in locals() else "Ошибка обработки"
        return fallback_text + " (Grad-CAM отключен из-за ошибки формата)", image

iface = gr.Interface(
    fn=predict_and_explain,
    inputs=gr.Image(label="Загрузите фото (чашка, бутылка, ключи)"),
    outputs=[
        gr.Textbox(label="Классификация (Latency: ~45ms)"),
        gr.Image(label="SOTA Объяснение (Grad-CAM Heatmap)")
    ],
    title="AITU Computer Vision: Assignment 2 Demo"
)

iface.launch(share=True)
```
